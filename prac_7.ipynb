```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
base_url = 'https://lifehacker.ru/'

```

```python
page_links = []
for page_num in range(1, 11):
    page_url = f"https://lifehacker.ru/topics/technology/?page={page_num}"
    response = requests.get(page_url)
    soup = BeautifulSoup(response.text, 'lxml')
    article_cards = soup.find_all('div', class_='lh-small-article-card')
    for card in article_cards:
    # Ищем ссылку в заголовке карточки
        title_link = card.find('a', class_='lh-small-article-card__link')
        if title_link and title_link.get('href'):
            href = title_link['href']
            if href.startswith('/'):
                full_url = base_url + href
            else:
                full_url = href
            page_links.append(full_url)
len(page_links)
```

```python
result = []
for url in page_links:
    article = {}
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'lxml')
    article['title'] = soup.find('title').text
    article_text = ""
    content_elem = soup.select_one('.single-article__content')
    if content_elem:
        paragraphs = content_elem.find_all('p')
        for p in paragraphs:
            text = p.get_text(strip=True)
            if len(text) > 20:
                article_text += text + "\n\n"
    article['content'] = article_text
    result.append(article)
```

```python
data = pd.DataFrame(result)
data.head()
```
